### Understanding Neural Networks
The [TensorFlow playground](https://playground.tensorflow.org/) is a handy neural network simulator built by the TensorFlow team. In this exercise, you will train several binary classifiers in just a few clicks, and tweak the model architecture and its hyperparameters to gain some intuition on how neural networks work and what their hyperparameters do. 

*The "default" neural network is the one that is available when you first visit the site:*
- Data: Circles (top left dataset)
- Ratio of training to test data: 50%
- Noise: 0
- Batch size 10
- Features: X1 and X2
- Layers: 2 hidden layers
  - 4 neurons in first hidden layer
  - 2 neurons in second hidden layer
- Learning rate: 0.03
- Activation: Tanh
- Regularization: None
- Regularization rate: 0
- Problem type: Classification

#### 3. **Activation functions.** 

Reset the model and change the learning rate back to 0.03.  Train the default network again.  

* Reset the model and change the activation function to ReLU.  How does the time to find a solution and the decision boundary compare to the Tanh activation?

*Your answer:*

ReLU converged in about the same time, maybe slightly faster. The main difference is the boundary looks more angular/blocky since ReLU is piecewise linear, but it still separates the classes fine.

* Reset the model again and now change the activation function to Sigmoid.  How does the Sigmoid activation function compare to ReLU and Tanh? 

*Your answer:*

Sigmoid was noticeably slower. It saturates for large inputs so gradients get tiny (vanishing gradient problem), which makes training sluggish. Tanh is basically a centered version of Sigmoid so it's a bit better, but ReLU is faster than both.

* Why do you think ReLU is generally the preferred activation function in modern neural networks? 

*Your answer:*

Mostly because it doesn't saturate for positive inputs, so gradients don't vanish in deep networks. It's also way cheaper to compute than Sigmoid or Tanh. The only real downside is the dying ReLU problem, but that's manageable.
