*Your answer:*

* Based on your observations, explain how adding hidden layers might impact the model's ability to classify more complex datasets. 


*Your answer:*

Adding hidden layers allows a neural network to construct increasingly complex, non-linear decision boundaries by composing simpler functions layer by layer. Each additional hidden layer gives the network the ability to form more abstract representations: early layers detect low-level patterns (like edges or linear separations), while deeper layers combine those patterns into high-level structures (like curves or spirals). For simple datasets like the circles problem, two layers are usually sufficient. For highly complex, non-linearly-separable datasets — such as the spiral — deeper networks are often necessary because shallow networks lack the representational capacity to express the required boundary shape, no matter how wide they are. Depth effectively allows the model to factor complex functions into manageable compositional steps.

*Your answer:*

At a learning rate of **0.003** (10× smaller), the model converges significantly more slowly — it typically requires around **1,000–1,500 epochs** before the decision boundary stabilizes, compared to roughly 200 epochs at 0.03. * Reset the model again and change the learning rate to 3.  Run the model for at least 2000 epochs.  Was the model able to find a solution?   Do you think the model will converge eventually if you just let it run for a long time?  Why or why not?

*Your answer:*

At a learning rate of **3**, the model fails to find a good solution even after 2,000+ epochs. The training loss oscillates wildly and the decision boundary is chaotic and unstable — it jumps erratically between configurations rather than settling into a sensible pattern. I do **not** think the model would converge even with unlimited time. When the learning rate is excessively large, each gradient update overshoots the minimum of the loss surface: instead of stepping toward the minimum, the weights leap past it and land on the other side (or even farther away). This causes the optimization to bounce around or diverge. A very high learning rate essentially prevents gradient descent from exploiting the local curvature of the loss surface, so convergence is not just slow — it is structurally prevented.

* What are the consequences of a learning rate that is set too high or too low?

*Your answer:*

**Too high:** The weight updates are so large that they repeatedly overshoot the loss minimum. Training loss oscillates or even increases over time, the decision boundary is unstable, and the model may never converge. In severe cases the gradients can explode (NaN weights).

**Too low:** Each update is very small, so the model creeps toward the minimum extremely slowly. Training requires many more epochs, increasing computational cost. The model may also get stuck in a shallow local minimum or on a saddle point, since the small step size lacks the momentum to escape. In practice, both extremes are harmful: an ideal learning rate is large enough to converge in a reasonable time but small enough to reliably reach (and stay near) a good minimum.

#### 3. **Activation functions.** 

Reset the model and change the learning rate back to 0.03.  Train the default network again.  

* Reset the model and change the activation function to ReLU.  How does the time to find a solution and the decision boundary compare to the Tanh activation?

*Your answer:*

With **ReLU** at learning rate 0.03, the model converges in roughly the same number of epochs as Tanh (around 150–300 epochs), and sometimes slightly faster on this simple dataset. The key visual difference is in the **decision boundary shape**: ReLU neurons output zero for negative inputs and are linear for positive inputs, so the boundary tends to be composed of **piecewise-linear segments** (straight edges) rather than the smooth curves produced by Tanh. Despite this angular appearance, ReLU still correctly separates the two classes and achieves similarly low test loss. The boundary looks a bit more blocky or polygonal than the Tanh boundary, but it is functionally equivalent for this problem.

* Reset the model again and now change the activation function to Sigmoid.  How does the Sigmoid activation function compare to ReLU and Tanh? 

*Your answer:*

**Sigmoid** produces smooth, S-shaped activations and also converges to a reasonable solution on the Circles dataset, but it tends to train **noticeably more slowly** than both Tanh and ReLU at the same learning rate. The decision boundary is smooth like Tanh, but the convergence can be sluggish because Sigmoid saturates (outputs very close to 0 or 1) for large positive or negative inputs, causing the gradients in those regions to be very small — this is the **vanishing gradient** problem. Compared to ReLU, Sigmoid is slower and more prone to gradient vanishing in deep networks. Compared to Tanh (which is a re-scaled Sigmoid centered at zero), Sigmoid is slower to converge because its output range (0 to 1) is not zero-centered, which can slow gradient updates.

* Why do you think ReLU is generally the preferred activation function in modern neural networks? 

*Your answer:*

ReLU is preferred for several compelling reasons:

1. **No vanishing gradients (for active neurons):** Unlike Sigmoid and Tanh, which saturate and produce near-zero gradients for large input magnitudes, ReLU has a constant gradient of 1 for all positive inputs. This means gradients flow back through deep networks without shrinking, making training of very deep architectures feasible.
2. **Computational efficiency:** ReLU is simply `max(0, x)` — extremely cheap to compute compared to the exponential operations required by Sigmoid and Tanh.
3. **Sparse activations:** ReLU outputs exactly zero for negative inputs, which means many neurons are inactive at once. This sparsity can be computationally efficient and may serve as a form of implicit regularization.
4. **Fast convergence in practice:** Empirically, ReLU networks often converge faster than Sigmoid or Tanh networks on a wide range of tasks, as observed even in this playground exercise.

The main downside of ReLU is the "dying ReLU" problem (neurons that always output zero and stop learning), but variants like Leaky ReLU and ELU address this.

#### 4. **Impact of Regularization.** 
Set the learning rate to 0.03 and use the ReLU activation function.  Also change the dataset to the "blobs" data (lower left) and increase the Noise to 40.  Create a large neural network with 6 hidden layers with 6, 7, 8, 7, 6, 5 neurons in each of the layers, respectively. 

* Let the model run for at least 1000 epochs.  What do you notice about the decision boundary? 

*Your answer:*

With **no regularization**, the large network (6, 7, 8, 7, 6, 5 neurons across 6 hidden layers) trained on noisy blobs data produces a highly **jagged and irregular decision boundary** after 1,000+ epochs. The boundary twists and contorts to capture individual noisy training points rather than the broad separation between the two blobs. While the training loss becomes very low, the test loss remains noticeably higher — a clear sign of **overfitting**. The network has memorized idiosyncratic noise in the training data rather than learning the true underlying structure.

* Now add L2 Regularization with a Regularization Rate of 0.01.  Again let the model run for at least 1000 epochs. How does the regularized model compare with the unregularized model? 

*Your answer:*

With **L2 regularization at rate 0.01**, the decision boundary becomes considerably **smoother and more regular**. The model no longer tries to wrap tightly around every noisy training point; instead, it learns a broad, cleaner boundary that reflects the actual blob separation. The training loss is slightly higher than without regularization (the penalty term adds to the loss), but the **test loss is lower**, indicating better generalization. L2 regularization achieves this by penalizing large weight magnitudes, which forces the network to distribute its reliance across many small weights rather than relying heavily on a few, effectively constraining model complexity.

* What happens if you increase the regularization rate too much (for example, to 0.3)? 

*Your answer:*

With a regularization rate of **0.3**, the penalty on the weights becomes so dominant that the model is **severely underfitted**. Most of the neuron weights are driven toward zero, which essentially disables large portions of the network. The decision boundary collapses into a nearly flat or trivially simple boundary that barely distinguishes the two classes — essentially, the model learns almost nothing. Both training and test losses remain high. This demonstrates that regularization, while powerful for controlling overfitting, must be tuned carefully: too much regularization prevents the model from learning the signal in the data at all.

- What happens if you add more neurons.  Train your model with 1 hidden layer and 8 neurons.  Comment on how you feel the model performed.  

*Your answer:*

Width alone is not enough here — depth is needed.

- Experiment with the number of hidden layers and number of neurons per layer.  Can you find a combination that is able to find a stable and suitable solution?  If so what was it? (*I was able to find a network that worked reasonably well.*)

*Your answer:*

The key was combining sufficient **width** (at least 6–8 neurons in early layers to capture rich features) with sufficient **depth** (at least 3–4 layers to build hierarchical representations of the spiral structure). Networks that were wide but shallow, or deep but narrow, consistently underperformed.

* Comment on your findings about how the number of hidden layers and the number of neurons contribute to the model's performance.

*Your answer:*

These experiments reveal that **both depth and width are essential components of model capacity**, but they serve different roles:

- **Width (neurons per layer)** determines how many features the network can represent at each level of abstraction. Too few neurons per layer creates a bottleneck that limits information flow regardless of depth.
- **Depth (number of layers)** enables hierarchical feature learning — each layer builds more abstract representations from the previous layer's outputs. For complex, non-linear datasets like the spiral, depth is critical because the problem requires composing many simpler functions to represent the full decision boundary.
- For simple problems like Circles, a shallow, moderately wide network suffices. For complex problems like the Spiral, neither extreme (all depth, no width; or all width, no depth) works well — you need both.
- There are also diminishing returns: beyond a certain size, adding more neurons or layers does not improve performance and may hurt it (more overfitting, harder optimization). The right architecture balances capacity with the complexity of the target function.
